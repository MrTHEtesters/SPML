```python
from PIL import Image
import numpy as np
```

```python
image = Image.open("airplane.png")
image = image.convert('RGB')
width, height = image.size

watermarked_image = image.copy()
image_array = np.array(image)
watermarked_image_array = np.array(watermarked_image)

for i in range(height):
    for j in range(width):
        if (i + j) % 2 == 0:
            # Red channel.
            watermarked_image_array[i, j, 0] = min(image_array[i, j, 0] + 1, 255)

new_img = Image.fromarray(watermarked_image_array)
new_img.save("watermarked_airplane.png")
```

```python
image = Image.open("airplane.png").convert('RGB')
watermarked_image = Image.open("watermarked_airplane.png").convert('RGB')

image_array = np.array(image)
watermarked_image_array = np.array(watermarked_image)


difference_array = np.abs(image_array - watermarked_image_array)

difference_image = Image.fromarray(difference_array.astype(np.uint8))
difference_image.save("difference.png")
```
